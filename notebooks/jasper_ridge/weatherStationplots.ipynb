```python
import pandas as pd
import matplotlib.pyplot as plt
import datetime
import numpy as np
```

Temp Out, Out Hum, Wind Speed, Bar, Dew Pt., Wind Dir

```python
df = pd.read_csv("../../external_data/jasper_ridge/Jasper Ridge Weather Station/JR_weather_Jan-Feb_2026_reformatted_N5_bounds.csv", parse_dates=[0])
print(df.columns)
```

```python
def plotWeatherStation(yAxis, opacity, log, ylim, ylabel):
    """
    Plot weather station data with respect to inputs

    Inputs:
        yAxis: str - What data to plot
        opacity: float - Opacity of the plotted data
        log: Boolean - Whether or not to plot in log scale
        ylim: int - Limit of the y-axis
        ylabel: str - Label of y-axis
    Outputs:
        None: Plots a weather station graph
    """
    _, ax = plt.subplots(1, 1, figsize=(10, 4)) #
    curr_data_df = df
    ax.plot(curr_data_df["Datetime"], curr_data_df[yAxis], alpha=opacity)
    
    if log:
        ax.set_yscale("log")

    earliest_day = curr_data_df['Datetime'].iloc[0].date()
    latest_day = curr_data_df['Datetime'].iloc[-1].date()

    sunrise = datetime.datetime.strptime("06:00:00", "%H:%M:%S").time() 
    sunset = datetime.datetime.strptime("18:00:00", "%H:%M:%S").time()

    # Print each day between the earliest and latest day
    num_days = (latest_day - earliest_day).days + 1
    for i in range(num_days):
        curr_day = earliest_day + datetime.timedelta(days=i)

        curr_sunrise_datetime = datetime.datetime.combine(curr_day, sunrise)
        curr_sunset_datetime = datetime.datetime.combine(curr_day, sunset)
        ax.axvspan(curr_sunset_datetime - datetime.timedelta(days=1), 
                curr_sunrise_datetime, 
                color='#bfbfbf', alpha=0.25, zorder=0)
    ax.set_xlabel("Time")
    ax.set_ylabel(ylabel)
    # ax.set_title(curr_data_df.iloc[0]["stationID"])
    plt.grid(True, c="#a8a8a8")
    plt.ylim(top = ylim)
    if log:
        plt.savefig(f"JROO_DWS_Log_{ylabel}.png", dpi = 150, bbox_inches="tight")
    else:
        plt.savefig(f"JROO_DWS_{ylabel}.png", dpi = 150, bbox_inches="tight")


```

```python
plotWeatherStation("Temp Out", 1, False, 25, "Temperature")
```

```python
plotWeatherStation("Out Hum", 1, False, None, "Relative Humidity")
```

```python
plotWeatherStation("Wind Speed", 1, False, None, "Wind Speed")
```

```python
plotWeatherStation(" Bar  ", 1, False, None, "Pressure")
```

```python
plotWeatherStation("Dew Pt.", 1, False, None, "Dew Point")
```

```python
def convertWindDir(windDir):
    cardinalToNumeric = {"N": complex(0, 1), "E": complex(1, 0), "S": complex(0, -1), "W": complex(-1,0)}
    numericWindDir = 0
    for cardinalDirection in windDir:
        numericWindDir += cardinalToNumeric[cardinalDirection]
    print(numericWindDir)
    return -(np.angle(numericWindDir) * 180 / np.pi - 90) %360
```
